# Interactive Visuals: La Chasse aux Trésors Cosmique

Welcome to the interactive sandbox for the K3 Dark Matter model! Run the cells below to visualize the physics.

```python
import numpy as np
import plotly.graph_objects as go
from ipywidgets import interact, widgets
from IPython.display import display

# 1. The 'Goldilocks' Galaxy Slider
def plot_stellar_stream(manifold_type):
    x = np.linspace(-10, 10, 100)
    if manifold_type == '6D Symmetry (Calabi-Yau)':
        # Violent explosion / scattering of stars due to heavy dark matter
        y = np.random.normal(0, 5, 100)
        title = '6D Symmetry: Stars Violently Explode (Disruption)'
        color = 'red'
    else:
        # Peaceful flow of stars due to lighter 4D K3 asymmetric dark matter
        y = np.sin(x) + np.random.normal(0, 0.2, 100)
        title = '4D K3 Asymmetry: Stars Flow Peacefully (Stable Stream)'
        color = 'blue'
        
    fig = go.Figure(data=go.Scatter(x=x, y=y, mode='markers', marker=dict(color=color)))
    fig.update_layout(title=title, xaxis_title='Position along stream', yaxis_title='Perpendicular deviation', height=400)
    fig.show()

print("Interactive: The Goldilocks Galaxy Slider")
interact(plot_stellar_stream, manifold_type=widgets.RadioButtons(options=['6D Symmetry (Calabi-Yau)', '4D K3 Asymmetry (S_1,2 / S_2,1)'], value='4D K3 Asymmetry (S_1,2 / S_2,1)', description='Manifold:'))
```

```python
# 2. The Chameleon Shield Button
def plot_chameleon_shield(shield_active):
    r = np.linspace(1, 10, 100)
    if shield_active:
        # Wave shrinks and is swallowed (Chameleon mechanism active)
        wave = np.exp(-r)*np.sin(10*r)
        title = 'Chameleon Shield ON: Wave Shrinks & Bounces / Swallowed'
        color = 'green'
    else:
        # Giant wave
        wave = np.sin(r)
        title = 'Chameleon Shield OFF: Giant Wave Extends (Drains M87* Spin)'
        color = 'orange'
        
    fig = go.Figure(data=go.Scatter(x=r, y=wave, line=dict(color=color, width=3)))
    fig.update_layout(title=title, xaxis_title='Distance from M87* Center / Shield', yaxis_title='Axion Wave Amplitude', height=400)
    fig.show()

print("Interactive: The Chameleon Shield Toggle")
interact(plot_chameleon_shield, shield_active=widgets.ToggleButton(value=False, description='Toggle Chameleon Shield', button_style='info', icon='check'))
```

```python
# 3. The 15-Day Hiding Trick
def plot_hiding_trick(days_to_show):
    t = np.linspace(0, days_to_show, int(days_to_show*10))
    
    # 14.76-day Lunar Tide (Blue)
    lunar = np.sin(2 * np.pi * t / 14.76)
    # 15.05-day Dark Matter wave (Gold)
    dm = np.sin(2 * np.pi * t / 15.05)
    
    fig = go.Figure()
    fig.add_trace(go.Scatter(x=t, y=lunar, mode='lines', name='14.76-day Lunar Tide', line=dict(color='blue', width=2)))
    fig.add_trace(go.Scatter(x=t, y=dm, mode='lines', name='15.05-day Dark Matter', line=dict(color='gold', width=2)))
    
    fig.update_layout(title='The Perfect Camouflage: Lunar Tide vs Dark Matter', xaxis_title='Days', yaxis_title='Amplitude', height=400)
    fig.show()

print("Interactive: The 15-Day Hiding Trick")
interact(plot_hiding_trick, days_to_show=widgets.IntSlider(min=15, max=365, step=15, value=30, description='Days:'))
```
